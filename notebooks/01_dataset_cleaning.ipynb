```python
from collections import Counter

all_user_ids = []
category_ids = {}

for path in user_files:
    df = pd.read_csv(path, usecols=["id"])

    category = path.relative_to(DATASET_DIR).parts[0].replace(".csv", "")
    ids = df["id"].dropna().astype("int64")

    category_ids[category] = set(ids)
    all_user_ids.extend(ids.tolist())

id_counts = Counter(all_user_ids)

duplicate_ids = {
    user_id: count
    for user_id, count in id_counts.items()
    if count > 1
}

print("Total account records:", len(all_user_ids))
print("Unique account IDs:", len(id_counts))
print("IDs appearing more than once:", len(duplicate_ids))

if duplicate_ids:
    print("\nExample duplicate IDs:")
    print(list(duplicate_ids.items())[:10])
else:
    print("\n✅ No account IDs overlap between categories.")
```

```python
import pandas as pd
from pathlib import Path

DATASET_DIR = Path("../data/raw/datasets_full.csv")

user_files = [
    path for path in DATASET_DIR.rglob("users.csv")
    if "__MACOSX" not in path.parts
]

all_accounts = []

for path in user_files:
    df = pd.read_csv(path)

    # Identify dataset category
    category = path.relative_to(DATASET_DIR).parts[0].replace(".csv", "")

    # Add category and binary label
    df["account_type"] = category
    df["label"] = 0 if category == "genuine_accounts" else 1

    all_accounts.append(df)

master_accounts = pd.concat(
    all_accounts,
    ignore_index=True,
    sort=False
)

print("Master dataset shape:", master_accounts.shape)

print("\nClass distribution:")
print(master_accounts["label"].value_counts())

print("\nAccount categories:")
print(master_accounts["account_type"].value_counts())
```

```python
import numpy as np
import pandas as pd

# Copy only the columns we need
behavior_features = master_accounts[
    [
        "id",
        "account_type",
        "label",
        "statuses_count",
        "followers_count",
        "friends_count",
        "favourites_count",
        "listed_count",
        "created_at",
        "updated"
    ]
].copy()

# Convert dates to consistent UTC datetime
behavior_features["created_at_dt"] = pd.to_datetime(
    behavior_features["created_at"],
    errors="coerce",
    utc=True
)

behavior_features["updated_dt"] = pd.to_datetime(
    behavior_features["updated"],
    errors="coerce",
    utc=True
)

# --------------------------------------------------
# 1. ACCOUNT AGE
# --------------------------------------------------

behavior_features["account_age_days"] = (
    behavior_features["updated_dt"] -
    behavior_features["created_at_dt"]
).dt.total_seconds() / (60 * 60 * 24)

# --------------------------------------------------
# 2. STATUSES PER DAY
# --------------------------------------------------

behavior_features["statuses_per_day"] = np.where(
    behavior_features["account_age_days"] > 0,
    behavior_features["statuses_count"] /
    behavior_features["account_age_days"],
    np.nan
)

# --------------------------------------------------
# 3. FOLLOWER / FRIEND RATIO
# --------------------------------------------------

behavior_features["followers_friends_ratio"] = np.where(
    behavior_features["friends_count"] > 0,
    behavior_features["followers_count"] /
    behavior_features["friends_count"],
    np.nan
)

# --------------------------------------------------
# 4. FRIEND / FOLLOWER RATIO
# --------------------------------------------------

behavior_features["friends_followers_ratio"] = np.where(
    behavior_features["followers_count"] > 0,
    behavior_features["friends_count"] /
    behavior_features["followers_count"],
    np.nan
)

# --------------------------------------------------
# 5. FAVORITES PER STATUS
# --------------------------------------------------

behavior_features["favorites_per_status"] = np.where(
    behavior_features["statuses_count"] > 0,
    behavior_features["favourites_count"] /
    behavior_features["statuses_count"],
    np.nan
)

# --------------------------------------------------
# 6. LOG-TRANSFORMED COUNT FEATURES
# --------------------------------------------------

count_features = [
    "statuses_count",
    "followers_count",
    "friends_count",
    "favourites_count",
    "listed_count"
]

for col in count_features:
    behavior_features[f"log_{col}"] = np.log1p(
        behavior_features[col]
    )

# Remove temporary datetime columns
behavior_features.drop(
    columns=["created_at_dt", "updated_dt"],
    inplace=True
)

print("Behavior feature shape:", behavior_features.shape)

print("\nColumns:")
for i, col in enumerate(behavior_features.columns, start=1):
    print(f"{i}. {col}")
```
