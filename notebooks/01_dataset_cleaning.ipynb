```python
import pandas as pd
from pathlib import Path

DATASET_DIR = Path("../data/raw/datasets_full.csv")

user_files = [
    path for path in DATASET_DIR.rglob("users.csv")
    if "__MACOSX" not in path.parts
]

column_sets = []

for path in user_files:
    df = pd.read_csv(path)
    column_sets.append(set(df.columns))

common_columns = set.intersection(*column_sets)

print("Common columns across ALL datasets:")
for i, col in enumerate(sorted(common_columns), start=1):
    print(f"{i}. {col}")

print("\nNumber of common columns:", len(common_columns))
```

```python
all_columns = set().union(*column_sets)

unique_columns = all_columns - common_columns

print("Columns not shared by every dataset:")
for i, col in enumerate(sorted(unique_columns), start=1):
    print(f"{i}. {col}")

print("\nNumber of non-common columns:", len(unique_columns))
```

```python
genuine_tweets_path = (
    DATASET_DIR
    / "genuine_accounts.csv"
    / "tweets.csv"
)

genuine_tweets = pd.read_csv(genuine_tweets_path)

print("Shape:", genuine_tweets.shape)
print("\nColumns:")
print(genuine_tweets.columns.tolist())
```

```python
genuine_tweets = pd.read_csv(
    genuine_tweets_path,
    encoding="latin-1"
)

print("Shape:", genuine_tweets.shape)
print("\nColumns:")
print(genuine_tweets.columns.tolist())
```

```python
print("Missing values:")
print(genuine_tweets.isnull().sum().sort_values(ascending=False).head(15))
```

```python
print("Duplicate tweet IDs:", genuine_tweets["id"].duplicated().sum())
print("Unique users:", genuine_tweets["user_id"].nunique())
```

```python
tweet_counts = genuine_tweets["user_id"].value_counts()

print("Users with tweets:", tweet_counts.size)
print("\nTweet count statistics:")
print(tweet_counts.describe())
```

```python
dataset_info = []

for path in user_files:
    df = pd.read_csv(path)

    relative_path = path.relative_to(DATASET_DIR)
    category = relative_path.parts[0].replace(".csv", "")

    dataset_info.append({
        "category": category,
        "users": len(df),
        "columns": len(df.columns),
        "path": str(relative_path)
    })

dataset_info = pd.DataFrame(dataset_info)

display(dataset_info)
```
