```python
import os
import pandas as pd
import numpy as np

project_path = os.path.dirname(os.getcwd())

behavior_path = os.path.join(
    project_path,
    "data",
    "processed",
    "behavior_features.csv"
)

behavior_df = pd.read_csv(behavior_path)

print("Behavioral dataset loaded successfully!")
print("Shape:", behavior_df.shape)
```

```python
behavior_features = [
    "statuses_count",
    "followers_count",
    "friends_count",
    "favourites_count",
    "listed_count",
    "account_age_days",
    "statuses_per_day",
    "followers_friends_ratio",
    "friends_followers_ratio",
    "favorites_per_status",
    "log_statuses_count",
    "log_followers_count",
    "log_friends_count",
    "log_favourites_count",
    "log_listed_count"
]

X = behavior_df[behavior_features].copy()
y = behavior_df["label"].copy()

print("Number of behavioral features:", len(behavior_features))
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nLabel distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTest distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to test data
X_test_scaled = scaler.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

behavior_nn = Sequential([
    Dense(64, activation="relu", input_shape=(15,)),
    Dropout(0.3),

    Dense(32, activation="relu"),
    Dropout(0.2),

    Dense(1, activation="sigmoid")
])

behavior_nn.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

behavior_nn.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = behavior_nn.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=30,
    batch_size=32,
    callbacks=[early_stopping],
    verbose=1
)

print("\nTraining completed!")
print("Epochs actually completed:", len(history.history["loss"]))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predict probabilities
y_test_prob = behavior_nn.predict(X_test_scaled)

# Convert probabilities to binary predictions
y_test_pred = (y_test_prob >= 0.5).astype(int).ravel()

# Calculate metrics
nn_accuracy = accuracy_score(y_test, y_test_pred)
nn_precision = precision_score(y_test, y_test_pred)
nn_recall = recall_score(y_test, y_test_pred)
nn_f1 = f1_score(y_test, y_test_pred)

print("Behavioral Neural Network Performance")
print("=" * 55)
print(f"Accuracy : {nn_accuracy:.4f}")
print(f"Precision: {nn_precision:.4f}")
print(f"Recall   : {nn_recall:.4f}")
print(f"F1-Score : {nn_f1:.4f}")

print("\nDetailed Classification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=["Genuine", "Bot"]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.unique(y_train)

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weight_dict = dict(zip(classes, class_weights))

print("Class weights:")
print(class_weight_dict)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Create a fresh model
behavior_nn_weighted = Sequential([
    Dense(64, activation="relu", input_shape=(15,)),
    Dropout(0.3),

    Dense(32, activation="relu"),
    Dropout(0.2),

    Dense(1, activation="sigmoid")
])

behavior_nn_weighted.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Early stopping
early_stopping_weighted = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

# Train with class weights
history_weighted = behavior_nn_weighted.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=30,
    batch_size=32,
    class_weight=class_weight_dict,
    callbacks=[early_stopping_weighted],
    verbose=1
)

print("\nTraining completed!")
print(
    "Epochs actually completed:",
    len(history_weighted.history["loss"])
)
```

```python
y_weighted_prob = behavior_nn_weighted.predict(X_test_scaled)
y_weighted_pred = (y_weighted_prob >= 0.5).astype(int).ravel()

print("Weighted Behavioral Neural Network")
print("=" * 55)
print(f"Accuracy : {accuracy_score(y_test, y_weighted_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_weighted_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_test, y_weighted_pred, zero_division=0):.4f}")
print(f"F1-Score : {f1_score(y_test, y_weighted_pred, zero_division=0):.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_weighted_pred))
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

improved_nn = Sequential([
    Dense(128, activation="relu", input_shape=(15,)),
    BatchNormalization(),
    Dropout(0.3),

    Dense(64, activation="relu"),
    BatchNormalization(),
    Dropout(0.25),

    Dense(32, activation="relu"),
    Dropout(0.15),

    Dense(1, activation="sigmoid")
])

improved_nn.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

improved_nn.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping_improved = EarlyStopping(
    monitor="val_loss",
    patience=7,
    restore_best_weights=True
)

history_improved = improved_nn.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping_improved],
    verbose=1
)

print("\nTraining completed!")
print(
    "Epochs actually completed:",
    len(history_improved.history["loss"])
)
```

```python
from sklearn.utils import resample
import pandas as pd
import numpy as np

# Combine scaled features with labels
train_balanced_df = pd.DataFrame(
    X_train_scaled,
    columns=behavior_features
)
train_balanced_df["label"] = y_train.to_numpy()

# Separate genuine and bot accounts
genuine_train = train_balanced_df[train_balanced_df["label"] == 0]
bot_train = train_balanced_df[train_balanced_df["label"] == 1]

# Randomly reduce bots to match the number of genuine accounts
bot_train_balanced = resample(
    bot_train,
    replace=False,
    n_samples=len(genuine_train),
    random_state=42
)

# Combine them
balanced_train_df = pd.concat(
    [genuine_train, bot_train_balanced],
    axis=0
).sample(frac=1, random_state=42).reset_index(drop=True)

# Separate X and y
X_balanced = balanced_train_df[behavior_features].values
y_balanced = balanced_train_df["label"].values

print("Original training distribution:")
print(y_train.value_counts())

print("\nBalanced training distribution:")
print(pd.Series(y_balanced).value_counts())

print("\nBalanced training shape:")
print(X_balanced.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

balanced_nn = Sequential([
    Dense(128, activation="relu", input_shape=(15,)),
    BatchNormalization(),
    Dropout(0.3),

    Dense(64, activation="relu"),
    BatchNormalization(),
    Dropout(0.25),

    Dense(32, activation="relu"),
    Dropout(0.15),

    Dense(1, activation="sigmoid")
])

balanced_nn.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

balanced_nn.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping_balanced = EarlyStopping(
    monitor="val_loss",
    patience=7,
    restore_best_weights=True
)

history_balanced = balanced_nn.fit(
    X_balanced,
    y_balanced,
    validation_split=0.20,
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping_balanced],
    verbose=1
)

print("\nTraining completed!")
print(
    "Epochs actually completed:",
    len(history_balanced.history["loss"])
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Create imputer
imputer = SimpleImputer(strategy="median")

# Fit ONLY on training data
X_train_imputed = imputer.fit_transform(X_train)

# Apply same imputer to test data
X_test_imputed = imputer.transform(X_test)

# Scale after imputation
nn_scaler = StandardScaler()

X_train_nn = nn_scaler.fit_transform(X_train_imputed)
X_test_nn = nn_scaler.transform(X_test_imputed)

print("Training shape:", X_train_nn.shape)
print("Test shape:", X_test_nn.shape)

print("\nNaNs after imputation:")
print("Training:", np.isnan(X_train_nn).sum())
print("Test:", np.isnan(X_test_nn).sum())
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

clean_nn = Sequential([
    Dense(128, activation="relu", input_shape=(X_train_nn.shape[1],)),
    BatchNormalization(),
    Dropout(0.25),

    Dense(64, activation="relu"),
    BatchNormalization(),
    Dropout(0.20),

    Dense(32, activation="relu"),
    Dropout(0.10),

    Dense(1, activation="sigmoid")
])

clean_nn.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

clean_nn.summary()
```

```python
early_stopping_clean = EarlyStopping(
    monitor="val_loss",
    patience=7,
    restore_best_weights=True
)

history_clean = clean_nn.fit(
    X_train_nn,
    y_train,
    validation_split=0.20,
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping_clean],
    verbose=1
)

print("\nTraining completed!")
print(
    "Epochs actually completed:",
    len(history_clean.history["loss"])
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predict probabilities
test_probs_clean = clean_nn.predict(
    X_test_nn,
    verbose=0
).ravel()

# Convert probabilities to class predictions
test_pred_clean = (test_probs_clean >= 0.5).astype(int)

print("Cleaned Behavioral Neural Network Performance")
print("=" * 55)

print(f"Accuracy : {accuracy_score(y_test, test_pred_clean):.4f}")
print(f"Precision: {precision_score(y_test, test_pred_clean):.4f}")
print(f"Recall   : {recall_score(y_test, test_pred_clean):.4f}")
print(f"F1-Score : {f1_score(y_test, test_pred_clean):.4f}")

print("\nDetailed Classification Report:")
print(
    classification_report(
        y_test,
        test_pred_clean,
        target_names=["Genuine", "Bot"]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, test_pred_clean))
```
