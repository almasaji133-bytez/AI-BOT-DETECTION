```python
import pandas as pd
import numpy as np

DATA_PATH = "../data/processed/behavior_features.csv"

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
print("\nColumns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate IDs:", df["id"].duplicated().sum())

print("\nClass distribution:")
print(df["label"].value_counts())

print("\nAccount types:")
print(df["account_type"].value_counts())
```

```python
numeric_features = [
    "statuses_count",
    "followers_count",
    "friends_count",
    "favourites_count",
    "listed_count",
    "account_age_days",
    "statuses_per_day",
    "followers_friends_ratio",
    "friends_followers_ratio",
    "favorites_per_status"
]

print("Feature comparison: Genuine vs Bot\n")

feature_comparison = df.groupby("label")[numeric_features].mean().T

feature_comparison.columns = ["Genuine", "Bot"]

print(feature_comparison)
```

```python
category_comparison = df.groupby("account_type")[numeric_features].mean().T

print(category_comparison)
```

```python
import matplotlib.pyplot as plt

features_to_plot = [
    "statuses_per_day",
    "followers_friends_ratio",
    "friends_followers_ratio",
    "favorites_per_status"
]

for feature in features_to_plot:
    plt.figure(figsize=(8, 5))

    df.boxplot(
        column=feature,
        by="label"
    )

    plt.title(f"{feature}: Genuine vs Bot")
    plt.suptitle("")
    plt.xlabel("Label (0 = Genuine, 1 = Bot)")
    plt.ylabel(feature)

    plt.show()
```
