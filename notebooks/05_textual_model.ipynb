```python
import pandas as pd
import os

# Store loaded tweet data
tweet_data = []

for relative_path in tweet_files:
    path = os.path.join(raw_path, relative_path)

    if not os.path.exists(path):
        continue

    print(f"\nLoading: {relative_path}")

    try:
        # Read only columns useful for textual analysis
        df_temp = pd.read_csv(
            path,
            encoding="latin1",
            low_memory=False,
            usecols=lambda col: col in [
                "text",
                "user_id",
                "created_at",
                "timestamp"
            ]
        )

        # Remove rows without text
        df_temp = df_temp.dropna(subset=["text"])

        # Identify account type from file path
        account_type = relative_path.split("/")[0].replace(".csv", "")

        df_temp["account_type"] = account_type

        tweet_data.append(df_temp)

        print("Tweets loaded:", len(df_temp))

    except Exception as e:
        print("ERROR:", type(e).__name__)
        print(e)

print("\n" + "="*70)
print("Files successfully loaded:", len(tweet_data))
```

```python
dataset_info = {
    "fake_followers": 1,
    "genuine_accounts": 0,
    "social_spambots_1": 1,
    "social_spambots_2": 1,
    "social_spambots_3": 1,
    "traditional_spambots_1": 1
}

print("Available textual datasets:")
for name, label in dataset_info.items():
    print(f"{name:25} → label {label}")
```

```python
# Combine tweets from the available datasets
# while keeping only the information needed for the textual model.

text_dfs = []

for df_temp in tweet_data:
    account_type = df_temp["account_type"].iloc[0]
    label = dataset_info[account_type]

    temp = df_temp[["user_id", "text"]].copy()
    temp["label"] = label
    temp["account_type"] = account_type

    # Remove missing/empty tweets
    temp["text"] = temp["text"].astype(str).str.strip()
    temp = temp[temp["text"].str.len() > 0]

    text_dfs.append(temp)

text_tweets = pd.concat(text_dfs, ignore_index=True)

print("Combined tweet dataset shape:", text_tweets.shape)
print("\nLabel distribution:")
print(text_tweets["label"].value_counts())

print("\nAccount type distribution:")
print(text_tweets["account_type"].value_counts())

print("\nUnique accounts:", text_tweets["user_id"].nunique())

display(text_tweets.head())
```

```python
account_labels = (
    text_tweets
    .groupby("user_id")
    .agg(
        label=("label", "first"),
        account_type=("account_type", "first"),
        tweet_count=("text", "count")
    )
    .reset_index()
)

print("Account-level dataset shape:", account_labels.shape)

print("\nAccount labels:")
print(account_labels["label"].value_counts())

print("\nAccount types:")
print(account_labels["account_type"].value_counts())

print("\nTweets per account:")
print(account_labels["tweet_count"].describe())
```

```python
# Normalize account IDs before matching

behavior_ids = set(
    behavior_df["id"]
    .astype("int64")
    .astype(str)
)

text_tweets["user_id"] = pd.to_numeric(
    text_tweets["user_id"],
    errors="coerce"
)

text_tweets = text_tweets.dropna(subset=["user_id"])

text_tweets["user_id"] = (
    text_tweets["user_id"]
    .astype("int64")
    .astype(str)
)

print("Behavioral accounts:", len(behavior_ids))
print("Textual accounts:", text_tweets["user_id"].nunique())

matched_ids = set(text_tweets["user_id"]) & behavior_ids

print("Matched accounts:", len(matched_ids))
print(
    "Match percentage:",
    round(len(matched_ids) / len(behavior_ids) * 100, 2),
    "%"
)
```

```python
# Create account-level textual dataset
# using behavioral labels as the ground truth

account_labels = (
    behavior_df[["id", "label", "account_type"]]
    .copy()
)

account_labels["id"] = (
    account_labels["id"]
    .astype("int64")
    .astype(str)
)

text_account_data = (
    text_tweets
    .groupby("user_id")
    .agg(
        tweet_count=("text", "count"),
        text=("text", lambda x: " ".join(x))
    )
    .reset_index()
)

# Match textual accounts with behavioral labels
text_account_data = text_account_data.merge(
    account_labels,
    left_on="user_id",
    right_on="id",
    how="inner"
)

print("Final account-level textual dataset:")
print(text_account_data.shape)

print("\nLabels:")
print(text_account_data["label"].value_counts())

print("\nAccount types:")
print(text_account_data["account_type"].value_counts())

print("\nTweets per account:")
print(text_account_data["tweet_count"].describe())
```

```python
# ============================================================
# CONTROLLED TWEET SAMPLING
# Maximum 100 tweets per account
# ============================================================

MAX_TWEETS_PER_ACCOUNT = 100

# Keep only the columns we need
controlled_tweets = text_tweets[
    ["user_id", "text", "label", "account_type"]
].copy()

# Shuffle tweets so accounts with many tweets don't always
# contribute only their earliest tweets
controlled_tweets = controlled_tweets.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Take at most 100 tweets per account
controlled_tweets = (
    controlled_tweets
    .groupby("user_id", group_keys=False)
    .head(MAX_TWEETS_PER_ACCOUNT)
    .reset_index(drop=True)
)

print("=" * 70)
print("CONTROLLED TEXTUAL DATASET")
print("=" * 70)

print("\nShape:", controlled_tweets.shape)

print("\nTweets per account:")
print(
    controlled_tweets
    .groupby("user_id")
    .size()
    .describe()
)

print("\nMaximum tweets per account:",
      controlled_tweets.groupby("user_id").size().max())

print("\nUnique accounts:",
      controlled_tweets["user_id"].nunique())

print("\nLabels:")
print(controlled_tweets["label"].value_counts())

print("\nAccount types:")
print(controlled_tweets["account_type"].value_counts())
```

```python
# ============================================================
# CREATE ACCOUNT-LEVEL TEXT REPRESENTATION
# ============================================================

account_text = (
    controlled_tweets
    .groupby("user_id")
    .agg(
        text=("text", lambda x: " ".join(x)),
        label=("label", "first"),
        account_type=("account_type", "first"),
        tweet_count_used=("text", "count")
    )
    .reset_index()
)

print("=" * 70)
print("FINAL CONTROLLED ACCOUNT-LEVEL TEXT DATASET")
print("=" * 70)

print("\nShape:", account_text.shape)

print("\nLabels:")
print(account_text["label"].value_counts())

print("\nAccount types:")
print(account_text["account_type"].value_counts())

print("\nTweets used per account:")
print(account_text["tweet_count_used"].describe())

display(account_text.head())
```

```python
from sklearn.model_selection import train_test_split

# ============================================================
# ACCOUNT-LEVEL TRAIN / VALIDATION / TEST SPLIT
# ============================================================

# First: 70% train, 30% temporary
train_text, temp_text = train_test_split(
    account_text,
    test_size=0.30,
    stratify=account_text["label"],
    random_state=42
)

# Second: split the temporary 30% into
# 15% validation and 15% test
val_text, test_text = train_test_split(
    temp_text,
    test_size=0.50,
    stratify=temp_text["label"],
    random_state=42
)

print("=" * 70)
print("ACCOUNT-LEVEL TEXT SPLIT")
print("=" * 70)

print("\nTrain shape:", train_text.shape)
print("Validation shape:", val_text.shape)
print("Test shape:", test_text.shape)

print("\nTrain labels:")
print(train_text["label"].value_counts())

print("\nValidation labels:")
print(val_text["label"].value_counts())

print("\nTest labels:")
print(test_text["label"].value_counts())

print("\nTotal accounts:",
      len(train_text) + len(val_text) + len(test_text))

# Verify that no account appears in multiple splits
train_ids = set(train_text["user_id"])
val_ids = set(val_text["user_id"])
test_ids = set(test_text["user_id"])

print("\nAccount overlap checks:")
print("Train ∩ Validation:", len(train_ids & val_ids))
print("Train ∩ Test:", len(train_ids & test_ids))
print("Validation ∩ Test:", len(val_ids & test_ids))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# ============================================================
# TF-IDF TEXT REPRESENTATION
# ============================================================

tfidf = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

# Fit ONLY on training accounts
X_train_tfidf = tfidf.fit_transform(train_text["text"])

# Transform validation and test using the SAME vectorizer
X_val_tfidf = tfidf.transform(val_text["text"])
X_test_tfidf = tfidf.transform(test_text["text"])

y_train_text = train_text["label"]
y_val_text = val_text["label"]
y_test_text = test_text["label"]

print("=" * 70)
print("TF-IDF REPRESENTATION")
print("=" * 70)

print("\nVocabulary size:", len(tfidf.vocabulary_))

print("\nTrain TF-IDF shape:", X_train_tfidf.shape)
print("Validation TF-IDF shape:", X_val_tfidf.shape)
print("Test TF-IDF shape:", X_test_tfidf.shape)

print("\nSparse matrix:")
print("Train:", type(X_train_tfidf))
print("Validation:", type(X_val_tfidf))
print("Test:", type(X_test_tfidf))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# TF-IDF + LOGISTIC REGRESSION
# ============================================================

text_lr = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

text_lr.fit(X_train_tfidf, y_train_text)

# Validation predictions
val_pred = text_lr.predict(X_val_tfidf)

print("=" * 70)
print("TEXTUAL BASELINE — TF-IDF + LOGISTIC REGRESSION")
print("=" * 70)

print("\nValidation Performance")
print("-" * 50)

print(f"Accuracy : {accuracy_score(y_val_text, val_pred):.4f}")
print(f"Precision: {precision_score(y_val_text, val_pred):.4f}")
print(f"Recall   : {recall_score(y_val_text, val_pred):.4f}")
print(f"F1-Score : {f1_score(y_val_text, val_pred):.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_val_text,
        val_pred,
        target_names=["Genuine", "Bot"]
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_val_text, val_pred))
```

```python
# ============================================================
# NEURAL TEXT MODEL — PREPARE ACCOUNT TEXT
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("=" * 70)
print("PREPARING TEXT FOR NEURAL MODEL")
print("=" * 70)

# Maximum vocabulary
MAX_WORDS = 30000

# Maximum sequence length
MAX_LEN = 300

# Tokenizer
tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

# IMPORTANT:
# Fit tokenizer ONLY on training accounts
tokenizer.fit_on_texts(train_text["text"])

# Convert text → integer sequences
X_train_seq = tokenizer.texts_to_sequences(train_text["text"])
X_val_seq = tokenizer.texts_to_sequences(val_text["text"])
X_test_seq = tokenizer.texts_to_sequences(test_text["text"])

# Pad sequences
X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_val_pad = pad_sequences(
    X_val_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

print("\nVocabulary size:", len(tokenizer.word_index))

print("\nTraining shape:", X_train_pad.shape)
print("Validation shape:", X_val_pad.shape)
print("Test shape:", X_test_pad.shape)

print("\nData type:", X_train_pad.dtype)
```

```python
# ============================================================
# NEURAL TEXT MODEL — BIDIRECTIONAL LSTM
# ============================================================

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    SpatialDropout1D,
    Bidirectional,
    LSTM,
    Dense,
    Dropout
)
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

print("=" * 70)
print("BUILDING BIDIRECTIONAL LSTM TEXT MODEL")
print("=" * 70)

# ------------------------------------------------------------
# Model parameters
# ------------------------------------------------------------

MAX_WORDS = 30000
EMBED_DIM = 128
LSTM_UNITS = 64

# ------------------------------------------------------------
# Build model
# ------------------------------------------------------------

text_nn = Sequential([
    
    Embedding(
        input_dim=MAX_WORDS,
        output_dim=EMBED_DIM,
        mask_zero=True
    ),

    SpatialDropout1D(0.25),

    Bidirectional(
        LSTM(
            LSTM_UNITS,
            dropout=0.20,
            recurrent_dropout=0.10
        )
    ),

    Dense(64, activation="relu"),

    Dropout(0.40),

    Dense(1, activation="sigmoid")
])

# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

text_nn.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

# ------------------------------------------------------------
# Display architecture
# ------------------------------------------------------------

text_nn.summary()

# ------------------------------------------------------------
# Class weights
# ------------------------------------------------------------

classes = np.unique(y_train_text)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train_text
)

class_weights = dict(
    zip(classes, weights)
)

print("\nClass weights:")
print(class_weights)

# ------------------------------------------------------------
# Callbacks
# ------------------------------------------------------------

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

print("\nModel ready.")
```

```python
# ============================================================
# TRAIN BIDIRECTIONAL LSTM
# ============================================================

print("=" * 70)
print("TRAINING BIDIRECTIONAL LSTM TEXT MODEL")
print("=" * 70)

history_text_nn = text_nn.fit(
    X_train_pad,
    y_train_text,
    
    validation_data=(
        X_val_pad,
        y_val_text
    ),
    
    epochs=20,
    batch_size=64,
    
    class_weight=class_weights,
    
    callbacks=[
        early_stopping,
        reduce_lr
    ],
    
    verbose=1
)

print("\n" + "=" * 70)
print("TEXTUAL NEURAL NETWORK TRAINING COMPLETED")
print("=" * 70)

print(
    "Epochs actually completed:",
    len(history_text_nn.history["loss"])
)
```

```python
# ============================================================
# TEXTUAL ROBUSTNESS — LEAVE-ONE-BOT-TYPE-OUT
# TF-IDF + LINEAR SVM
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("=" * 70)
print("TEXTUAL ROBUSTNESS — LEAVE-ONE-BOT-TYPE-OUT")
print("=" * 70)

# Bot types that can be held out
bot_types = [
    "fake_followers",
    "social_spambots_1",
    "social_spambots_2",
    "social_spambots_3",
    "traditional_spambots_1"
]

robustness_results = []

for held_out in bot_types:

    print("\n" + "=" * 70)
    print("HELD-OUT BOT TYPE:", held_out)
    print("=" * 70)

    # --------------------------------------------------------
    # TRAINING DATA
    # Genuine accounts + all bot types EXCEPT held-out type
    # --------------------------------------------------------

    train_robust = account_text[
        (account_text["account_type"] == "genuine_accounts") |
        (
            (account_text["label"] == 1) &
            (account_text["account_type"] != held_out)
        )
    ].copy()

    # --------------------------------------------------------
    # TEST DATA
    # Genuine accounts + ONLY held-out bot type
    # --------------------------------------------------------

    test_robust = account_text[
        (account_text["account_type"] == "genuine_accounts") |
        (account_text["account_type"] == held_out)
    ].copy()

    print("Training accounts:", len(train_robust))
    print("Testing accounts :", len(test_robust))

    print("\nTraining distribution:")
    print(train_robust["account_type"].value_counts())

    print("\nTesting distribution:")
    print(test_robust["account_type"].value_counts())

    # --------------------------------------------------------
    # TF-IDF
    # Fit ONLY on training data
    # --------------------------------------------------------

    vectorizer = TfidfVectorizer(
        max_features=20000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        sublinear_tf=True
    )

    X_train_robust = vectorizer.fit_transform(
        train_robust["text"]
    )

    X_test_robust = vectorizer.transform(
        test_robust["text"]
    )

    y_train_robust = train_robust["label"].values
    y_test_robust = test_robust["label"].values

    # --------------------------------------------------------
    # Linear SVM
    # --------------------------------------------------------

    svm_robust = LinearSVC(
        class_weight="balanced",
        random_state=42
    )

    svm_robust.fit(
        X_train_robust,
        y_train_robust
    )

    # --------------------------------------------------------
    # Predictions
    # --------------------------------------------------------

    test_pred = svm_robust.predict(
        X_test_robust
    )

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test_robust,
        test_pred
    )

    precision = precision_score(
        y_test_robust,
        test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test_robust,
        test_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test_robust,
        test_pred,
        zero_division=0
    )

    print("\nPerformance on UNSEEN bot type:")
    print("-" * 50)

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-Score : {f1:.4f}")

    # --------------------------------------------------------
    # Save result
    # --------------------------------------------------------

    robustness_results.append({
        "held_out_type": held_out,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })


# ============================================================
# FINAL RESULTS TABLE
# ============================================================

robustness_df = pd.DataFrame(
    robustness_results
)

print("\n" + "=" * 70)
print("FINAL TEXTUAL ROBUSTNESS RESULTS")
print("=" * 70)

display(
    robustness_df.style.format({
        "accuracy": "{:.4f}",
        "precision": "{:.4f}",
        "recall": "{:.4f}",
        "f1": "{:.4f}"
    })
)

print("\nMean Accuracy:",
      robustness_df["accuracy"].mean())

print("Mean F1:",
      robustness_df["f1"].mean())
```
